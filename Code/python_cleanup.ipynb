## Dataset cleanup with Pandas and Python

### Step 1: Data exploration and cleanup

```python
# Import dependencies

import pandas as pd

df = pd.read_csv("../Data/Walmart.csv", encoding_errors="ignore")
df.head()
```

```python
df.describe()
df.info()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)

df.isnull().sum()
```

```python
df["unit_price"] = df["unit_price"].str.replace("$", "").astype(float)
df.info()
```

```python
df.insert(
    df.columns.get_loc("quantity") + 1,
    "total_revenue",
    df["unit_price"] * df["quantity"]
)

df.head()
```

```python
from pathlib import Path

output_path = Path.home() / "Downloads" / "walmart_cleaned.csv"

df.to_csv(output_path, index=False)

print(f"Saved to: {output_path}")
```
